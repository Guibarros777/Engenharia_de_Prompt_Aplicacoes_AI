```python
def verificar_temperatura(temperatura_atual):
    """
    Verifica a temperatura do sistema e exibe um alerta se for maior que 30 graus Celsius.
    Parâmetros:
    temperatura_atual (int/float): A temperatura atual do sistema a ser verificada.
    """
    # Verifica se a temperatura atual é superior a 30 graus Celsius
    if temperatura_atual > 30:
        # Se a temperatura for maior que 30, exibe um alerta de superaquecimento usando concatenação
        print("ALERTA: Superaquecimento! Temperatura atual: " + str(temperatura_atual) + "°C")
    else:
        # Se a temperatura for 30 ou menor, informa que a temperatura está normal usando concatenação
        print("Temperatura normal. Temperatura atual: " + str(temperatura_atual) + "°C")

# --- Solicitação de temperatura ao usuário ---

# Solicita ao usuário que insira a temperatura atual do sistema
try:
    temperatura_digitada = float(input("Por favor, digite a temperatura atual do sistema em Celsius: "))
    # Chama a função para verificar e imprimir o status da temperatura digitada pelo usuário
    verificar_temperatura(temperatura_digitada)
except ValueError:
    print("Entrada inválida. Por favor, digite um número para a temperatura.")
```

crie um progama que organize 10 arquivos bagunçados em pastas automaticamete,baseando-se apenas na extensão do arquivo,usando a biblioteca os ou shu
comente execução durante o codigo.(ex:.pdf,.jpg)


```python
import os
import shutil

def organizar_arquivos_por_extensao(diretorio_origem):
    """
    Organiza arquivos em um diretório criando pastas baseadas na extensão do arquivo.

    Parâmetros:
    diretorio_origem (str): O caminho do diretório contendo os arquivos a serem organizados.
    """
    if not os.path.exists(diretorio_origem):
        print(f"Erro: O diretório de origem '{diretorio_origem}' não existe.")
        return

    print(f"Iniciando a organização de arquivos em: '{diretorio_origem}'")

    arquivos_organizados = 0
    arquivos_nao_organizados = 0

    for filename in os.listdir(diretorio_origem):
        caminho_completo_origem = os.path.join(diretorio_origem, filename);

        # Ignora diretórios existentes no diretório de origem
        if os.path.isdir(caminho_completo_origem):
            print(f"Ignorando diretório: '{filename}'");
            continue

        # Separa o nome do arquivo da extensão
        nome_base, extensao = os.path.splitext(filename);
        # Remove o ponto da extensão e converte para minúsculas para padronização
        extensao = extensao[1:].lower();

        # Define a pasta de destino com base na extensão do arquivo
        if not extensao:
            pasta_destino_nome = 'sem_extensao' # Para arquivos sem extensão
        elif extensao in ['jpg', 'png', 'gif', 'jpeg']:
            pasta_destino_nome = 'imagens'
        elif extensao in ['pdf', 'docx', 'txt', 'xlsx', 'pptx']:
            pasta_destino_nome = 'documentos'
        elif extensao in ['py', 'js', 'html', 'css']:
            pasta_destino_nome = 'codigos'
        else:
            pasta_destino_nome = 'outros' # Para qualquer outra extensão não categorizada

        pasta_destino_caminho = os.path.join(diretorio_origem, pasta_destino_nome);

        # Cria a pasta de destino se ela não existir
        if not os.path.exists(pasta_destino_caminho):
            os.makedirs(pasta_destino_caminho);
            print(f"  Pasta de destino criada: '{pasta_destino_nome}/'");

        # Caminho completo do arquivo no destino
        caminho_completo_destino = os.path.join(pasta_destino_caminho, filename);

        # Move o arquivo para a pasta de destino
        try:
            shutil.move(caminho_completo_origem, caminho_completo_destino);
            print(f"  Movido '{filename}' para '{pasta_destino_nome}/'");
            arquivos_organizados += 1;
        except Exception as e:
            print(f"  Erro ao mover '{filename}': {e}");
            arquivos_nao_organizados += 1;

    print(f"\nOrganização concluída. {arquivos_organizados} arquivo(s) organizado(s), {arquivos_nao_organizados} arquivo(s) não organizado(s).");


# --- Exemplo de uso: Criar 10 arquivos de teste e depois organizá-los ---
# Define o diretório onde os arquivos de teste serão criados e organizados
diretorio_de_teste = 'arquivos_para_organizar_exemplo';

# Garante que o diretório de teste exista e esteja vazio para cada execução
# Isso remove o diretório e seu conteúdo se já existir, e então o cria novamente
if os.path.exists(diretorio_de_teste):
    shutil.rmtree(diretorio_de_teste);
os.makedirs(diretorio_de_teste, exist_ok=True);

print(f"Criando 10 arquivos de exemplo em '{diretorio_de_teste}' para demonstração...");

# Lista de 10 arquivos de exemplo com extensões variadas para simular arquivos bagunçados
arquivos_exemplo_para_criar = [
    'documento_importante.pdf',
    'foto_de_ferias.jpg',
    'codigo_fonte.py',
    'relatorio_final.docx',
    'planilha_gastos.xlsx',
    'musica_favorita.mp3',
    'imagem_logo.png',
    'backup.zip',
    'anotacoes.txt',
    'apresentacao.pptx'
];

for arquivo in arquivos_exemplo_para_criar:
    caminho_arquivo_teste = os.path.join(diretorio_de_teste, arquivo);
    with open(caminho_arquivo_teste, 'w') as f:
        f.write(f'Conteúdo de teste para o arquivo: {arquivo}');
print("Todos os 10 arquivos de exemplo foram criados com sucesso.");
print("-" * 60);

# Chama a função principal para organizar os arquivos no diretório de teste
organizar_arquivos_por_extensao(diretorio_de_teste);

print("-" * 60);
print(f"Verifique o diretório '{diretorio_de_teste}' para ver os resultados da organização.");

# Nota: Para limpar os arquivos de teste e as pastas criadas, você pode descomentar as linhas abaixo:
# print("\nLimpando arquivos de teste...");
# shutil.rmtree(diretorio_de_teste);
# print(f"Diretório de teste '{diretorio_de_teste}' e seus conteúdos foram removidos.");
```

CONSULTA SIMPLES DE API


```python
import requests # Importa a biblioteca requests para fazer requisições HTTP

def consultar_cep(cep):
    """
    Consulta um CEP utilizando a API ViaCEP e retorna informações de endereço.

    Parâmetros:
    cep (str): O CEP a ser consultado, pode conter hífens e pontos.

    Retorna:
    str: Uma string formatada com logradouro, bairro e cidade, ou uma mensagem de erro.
    """
    # Limpa o CEP removendo hífens, pontos e espaços em branco
    cep = cep.replace("-", "").replace(".", "").strip()

    # Verifica se o CEP tem 8 dígitos após a limpeza
    if len(cep) != 8:
        return "CEP inválido."

    # Monta a URL da API ViaCEP com o CEP fornecido
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        # Faz a requisição GET para a API ViaCEP
        response = requests.get(url)

        # Levanta um erro HTTP se a requisição não foi bem-sucedida (status code 4xx ou 5xx)
        response.raise_for_status()

        # Converte a resposta JSON para um dicionário Python
        dados = response.json()

        # Verifica se a resposta JSON contém a chave 'erro', indicando que o CEP não foi encontrado
        if "erro" in dados:
            return "CEP não encontrado na base de dados."

        # Retorna o logradouro, bairro e cidade formatados
        return f"Logradouro: {dados['logradouro']}, Bairro: {dados['bairro']}, Cidade: {dados['localidade']}"

    except requests.exceptions.RequestException as e:
        # Captura exceções relacionadas a problemas de conexão (rede, DNS, etc.)
        return f"Erro de conexão: {e}"

# --- Exemplo de uso da função ---

# Solicita ao usuário que digite o CEP
cep_digitado = input("Digite o CEP que deseja consultar: ")
# Chama a função de consulta e imprime o resultado
print(consultar_cep(cep_digitado))
```

sistema de notificação


```python
import datetime # Importa o módulo datetime para trabalhar com datas e horas.

def enviar_notificacao(mensagem, tipo="INFO"): # Define a função 'enviar_notificacao' que aceita uma mensagem e um tipo de notificação (padrão é 'INFO').
    """
    Envia uma notificação com uma mensagem e um tipo (INFO, ALERTA, ERRO).
    Em um cenário real, esta função poderia enviar e-mails, SMS, ou usar APIs de serviços de notificação.
    """
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S") # Obtém a data e hora atuais e formata como uma string.
    if tipo.upper() == "ALERTA": # Verifica se o tipo de notificação é 'ALERTA' (ignorando maiúsculas/minúsculas).
        print(f"[{timestamp}] [ALERTA] {mensagem}") # Imprime a notificação de alerta com timestamp.
    elif tipo.upper() == "ERRO": # Se não for 'ALERTA', verifica se é 'ERRO'.
        print(f"[{timestamp}] [ERRO] {mensagem}") # Imprime a notificação de erro com timestamp.
    else: # Se não for 'ALERTA' nem 'ERRO', assume que é 'INFO' ou outro tipo padrão.
        print(f"[{timestamp}] [INFO] {mensagem}") # Imprime a notificação de informação com timestamp.

# --- Exemplos de Uso ---
print("\n--- Testando o Sistema de Notificação ---") # Imprime um cabeçalho para a seção de testes.

# Notificação de informação
enviar_notificacao("Operação concluída com sucesso.", "INFO") # Chama a função para enviar uma notificação de informação.

# Notificação de alerta
enviar_notificacao("Atenção: Espaço em disco baixo.", "ALERTA") # Chama a função para enviar uma notificação de alerta.

# Notificação de erro
enviar_notificacao("Erro crítico: Conexão com o banco de dados falhou.", "ERRO") # Chama a função para enviar uma notificação de erro.

# Exemplo de uso com uma condição (similar ao sistema de temperatura)
def verificar_status_servico(status_ok): # Define uma função para verificar o status de um serviço com base em um booleano.
    if status_ok: # Se o status for verdadeiro (True).
        enviar_notificacao("Serviço principal online.", "INFO") # Envia uma notificação de informação.
    else: # Se o status for falso (False).
        enviar_notificacao("Serviço principal offline! Investigar imediatamente.", "ERRO") # Envia uma notificação de erro.

print("\n--- Verificação de Status do Serviço ---") # Imprime um cabeçalho para a verificação de status.
verificar_status_servico(True) # Testa a função com status OK.
verificar_status_servico(False) # Testa a função com status não OK.

print("Sistema de notificação pronto.") # Imprime uma mensagem indicando o fim da execução.
```
